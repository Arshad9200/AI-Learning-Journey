# Sorting data - sort by values

```python
# Ascending Order
import pandas as pd
data = {"Name": ["Arshad", "Aatif", "Abhijeet"], "Age": [26, 24, 25], "Marks": [92, 91, 99]}
df = pd.DataFrame(data)
print(df.sort_values("Marks"))

# Descending Order
print(df.sort_values("Marks", ascending=False))
```

# MERGE TWO OR MORE DATA SET / DATA FRAME - when columns are same

```python
import pandas as pd
df1 = pd.DataFrame({
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Age": [26, 23, 24, 25],
    "Salary": [24000, 28000, 18000, 16000],
    "Department": ["Support", "AI Developer", "Networking", "Cloud"],
    "Experience": [1.4, 2, 1.2, 1]
})
print(df1)
print()

df2 = pd.DataFrame({
    "Name": ["Manu", "Rashi", "Taukir", "Sid"],
    "Age": [27, 25, 24, 26],
    "Salary": [12000, 22000, 18000, 14000],
    "Department": ["Process", "Marketing", "Cloud", "Business"],
    "Experience": [0, 1.4, 1.3, 2]
})
print(df2)
print()

# Merge both data frame
final_list = pd.concat([df1, df2])
print(final_list)
```

# MERGE TWO OR MORE DATA SET / DATA FRAME - when columns are not same

```python
import pandas as pd
df3 = pd.DataFrame({
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Age": [26, 23, 24, 25],
})
print(df3)
print()

df4 = pd.DataFrame({
    "Name": ["Aatif", "Arshad", "Taukir", "Sid"],
    "Salary": [12000, 22000, 18000, 14000],
})
print(df4)
print()

# Merge both data frame
final_list1 = pd.merge(df3, df4, on="Name")
print(final_list1)
```
