# MATHEMATICAL OPERATION

```python
# Apply Matheematical opertaion by single element

import pandas as pd
series_data = pd.Series([12, 34, 43], name = "Marks of students")
# print(series_data)

# Addition single number in each
print("Add 5.5 in each numbers\n", series_data + 5.5)

# Subtract single number in each
print("Subtract 1 from each numbers\n", series_data - 1)

#  Multiply number in each
print("Multiply by 2 in each numbers\n", series_data * 2)

# Divide single number in each
print("Divide by 5 in each numbers\n", series_data / 5)
```

```python
# Apply Mathematical opertaion between two series

import pandas as pd
data1 = pd.Series([12, 34, 43])

data2 = pd.Series([12, 34, 43])

# Addition single number in each
print("Add both series\n", data1 + data2)

# Subtract single number in each
print("Subtract  both series\n", data1 - data2)

#  Multiply number in each
print("Multiply  both series\n", data1 * data2)

# Divide single number in each
print("Divide  both series\n", data1 / data2)
```

# USEFUL STATISTICAL METHODS

```python
import pandas as pd
series = pd.Series([12, 34, 43])
print(series)
print()

# SUM METHOD
print("Sum:", series.sum())

# MEAN(Average) METHOD
print("Mean:", series.mean())

# MINIMUM METHOD
print("Min:", series.min())

# MAXIMUM METHOD
print("Max:", series.max())

# COUNT METHOD
print("Count:", series.count())

# STD METHOD - How speard from mean (acg)
print("STD:", series.std())

# MEDIAN METHOD - Middle value
print("Median:", series.median())
```

# FILTERING DATA

```python
# Single filter condition
import pandas as pd
data = {
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Salary": [24000, 29000, 18000, 18000],
    "Dept": ["Saas", "AI/ML", "Cloud", "AWS"]
}
df = pd.DataFrame(data, index=["A", "B", "C", "D"])
print(df)
print()
print("Employees below 25k salary:", df[df["Salary"] >25000])
```

```python
# Multiple filter condition
import pandas as pd
data = {
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Salary": [24000, 29000, 18000, 18000],
    "Dept": ["Saas", "AI/ML", "Cloud", "AWS"]
}
df = pd.DataFrame(data, index=["A", "B", "C", "D"])
# print(df)
print("Fetch employees whose salary is more than 25k", df[df["Salary"] >= 24000])
print()
print("Fetch employee whose salary is <20k:\n",df[(df["Salary"] < 30000) | (df["Salary"] > 20000)])
```

# BASIC STATISTICAL OPERATIONS

```python
import pandas as pd
data = {
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Salary": [24000, 29000, 18000, 18000],
    "Dept": ["Saas", "AI/ML", "Cloud", "AWS"]
}
df = pd.DataFrame(data, index=["A", "B", "C", "D"])
print(df)
print()

# SUM METHOD
print("Sum:", df["Salary"].sum())

# MEAN METHOD
print("Mean(Avg):", df["Salary"].mean())

# MIN METHOD
print("Min:", df["Salary"].min())

# MAX METHOD
print("Max:", df["Salary"].max())

# MEDIAN METHOD
print("Median:", df["Salary"].median())

# STD METHOD
print("STR:", df["Salary"].std())

# COUNT METHOD
print("Count:", df["Salary"].count())

# DESCRIBE METHOD
print("Describe:\n", df["Salary"].describe())
```

# SORTING

```python
import pandas as pd
data = {
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Salary": [24000, 29000, 18000, 18000],
    "Dept": ["Saas", "AI/ML", "Cloud", "AWS"]
}
df = pd.DataFrame(data, index=["A", "B", "C", "D"])
print(df)
print()

# Ascending Order - default
print("Sort based on salary (Ascending Order)\n", df.sort_values("Salary"))
print()

# Descending Order 
print("Sort based on salary (Descending Order)\n", df.sort_values("Salary", ascending=False))
```

# MISSING VALUES

# INSERT 0 AT MISSING VALUE POSITION

```python
# Insert 0 in single column
import pandas as pd
import numpy as np
data = {
    "Name": ["Arshad", "Atif", "Abhijeet", "Ashu"],
    "Salary": [24000, 29000, np.nan, 18000],
    "Dept": ["Saas", "AI/ML", np.nan, np.nan]
}
df = pd.DataFrame(data, index=["A", "B", "C", "D"])
print(df)
print()
df["Dept"] = df["Dept"].fillna(0)
df["Salary"] = df["Salary"].fillna(0)
print("Insert 0 at columns containg Null Values\n", df)
```

# CONCAT DATAFRAMES

```python
# Concat method: Concat two dataframes
import pandas as pd
df1 = pd.DataFrame({
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Age": [26, 23, 24, 25],
    "Salary": [24000, 28000, 18000, 16000],
    "Department": ["Support", "AI Developer", "Networking", "Cloud"],
    "Experience": [1.4, 2, 1.2, 1]
})
df2 = pd.DataFrame({
    "Name": ["Om", "xyz", "Abhijeet", "Ashu"],
    "Age": [26, 23, 24, 25],
    "Salary": [24000, 28000, 18000, 16000],
    "Department": ["Support", "AI Developer", "Networking", "Cloud"],
    "Experience": [1.4, 2, 1.2, 1]
})
res = pd.concat([df1, df2])
print(res)
```

```python
# Merge method: merge two dataframes based on common column
import pandas as pd
df1 = pd.DataFrame({
    "Name": ["Arshad", "Aatif", "Abhijeet", "Ashu"],
    "Age": [26, 23, 24, 25],
    "Salary": [24000, 28000, 18000, 16000],
    "Department": ["Support", "AI Developer", "Networking", "Cloud"],
    "Experience": [1.4, 2, 1.2, 1]
})
df2 = pd.DataFrame({
    "Name": ["Om", "xyz", "Abhijeet", "Ashu"],
    "Salary": [24000, 28000, 18000, 16000],
})
res = pd.merge(df1, df2, on="Name")
print(res)
```
