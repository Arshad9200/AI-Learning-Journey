# SLICING

```python
# 1-D ARRAY
import numpy as np
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
print("Dimension:",arr1.ndim)
print("Shape:",arr1.shape)

fetch_2_to_5_element = arr1[1:5:1]
print("Fetch 2 to 5 elments:", fetch_2_to_5_element)


# Slicing with Start Only
fetch_all_from_3 = arr1[2:]
print("3 to all:", fetch_all_from_3)

# Slicing with Stop Only
stop_on_4 = arr1[:4]
print("stop on 4:", stop_on_4)


# Slicing the entire array
entire_array = arr1[:]
print("Entire Array:", entire_array)

# Slicing with steps
alternate_elements = arr1[::2]
print("Alternate Elements:", alternate_elements)

# Negetive slicing
negetive_slicing = arr1[-4:-1]
print("Negetive Slicing:", negetive_slicing)

# Reverse Array
reverse_array = arr1[::-1]
print("Reverse Array:", reverse_array)
```

```python
# Create Numpy Array  -  2D
import numpy as np
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60],
                [70, 80, 90]])
print(arr2)
print("Dimension:",arr2.ndim)
print("Shape:",arr2.shape)

# Get the First Row
first_row = arr2[0, :]
print("First Row:", first_row)

# Get the Specific element of second Row
sec_arr_elements = arr2[1, 1:3]
print("Second Array Elements:", sec_arr_elements)

# Get the First Column
last_column = arr2[:, 2]
print("Last Column:", last_column)

# Get the Specific element of third Columns
sec_third_col_elements = arr2[1:3, 2:3]
print("Second Third Specific Elements:", sec_third_col_elements)

# Get a Subarray
sub_array_50_60_70_80 = arr2[0:3, 1:3]
print("Sub Array:", sub_array_50_60_70_80)
```

# COPY VS VIEW FUNCTIONS

```python
# COPY
import numpy as np
original_arr = np.array([1, 2, 3, 4])
copy = original_arr.copy()
copy[0] = 100
print("Original Array: ",original_arr)
print("Copy:",copy)
```

```python
# VIEW
import numpy as np
original_arr = np.array([1, 2, 3, 4])
copy = original_arr.view()
copy[0] = 100
print("Original Array: ",original_arr)
print("Copy:",copy)
```

How to Check Whether Memory is Shared using Base method

```python
# COPY
import numpy as np
original_arr = np.array([1, 2, 3, 4])
copy = original_arr.copy()
print(original_arr.base)
# None means copy variable has its own memory
```

```python
# VIEW
import numpy as np
original_arr = np.array([1, 2, 3, 4])
copy = original_arr.view()
print(copy.base)
# [1 2 3 4]  means copy variable using the memory of the original_arr
```

# Where Function

```python
# 1D Numpy Array
import numpy as np
arr = np.array([1, 2, 3, 4, 4, 7, 4, 2, 1, 3])
print(arr)
result = np.where(arr == 4)
print("4 is available at index :", result)

# Find the even number index
even = np.where(arr % 2 == 0)
print("Index of even number:", even)

# Find the odd number index
odd = np.where(arr % 2 != 0)
print("Index of odd number:", odd)

# Find the greater than 3 numbers index
greater_than_2 = np.where(arr > 3)
print("Index of greater than 2 :", greater_than_2)
```

```python
import numpy as np
matrix1 = np.matrix([
    [10, 20, 30, 40],
    [50, 60, 70, 80]
])
matrix2 = np.matrix([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])
add_res = matrix1 - matrix2
print(add_res)
print()


# Subtraction in Numpy array
import numpy as np
matrix11 = np.matrix([
    [10, 20, 30, 40],
    [50, 60, 70, 80]
])
matrix12 = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])
add_res2 = matrix11 - matrix12
print(add_res2)
print()
```

# INVERSE FUNCTION

```python
# using np.linalg.inv() method
import numpy as np
arr = np.array([
    [4, 7],
    [2, 6]
])
print(arr)
print("Shape:",arr.shape)
print()
inverse_arr = np.linalg.inv(arr)
print(inverse_arr)

# Varify the inverse by multiplying arr * inverse_arr = identity matrix
varify_res = np.dot(arr, inverse_arr)
print(varify_res)
# Result is not identity matrix so that is not inverse
```

# POWER FUNCTION

```python
# using np.linalg.matrix_power() method - it is doing matrix multiplication not simple multiply
import numpy as np
arr = np.array([
    [4, 7],
    [2, 6]
])
print(arr)
print("Power 2: square matrix")
power2_arr = np.linalg.matrix_power(arr, 2)
print(power2_arr)
print("Power negetive: inverse matrix")
power_negetive_arr = np.linalg.matrix_power(arr, -1)
print(power_negetive_arr)
print("Power 0: Identity matrix")
power_zero_arr = np.linalg.matrix_power(arr, 0)
print(power_zero_arr)
```
