# LINE GRAPH

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(x, y)
plt.show()
```

```python
# Add Title
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(x, y)
plt.title("Sales Report")
plt.show()
```

```python
# Add X-Axis & Y-Axis Labels (axis name)
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.plot(month, sales)
plt.title("Monthly Sales Report")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# Change the Line Color
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.plot(month, sales, color="Blue")
plt.show()
```

```python
# Add Marker
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.plot(month, sales, marker="s")
plt.show()
```

```python
# Add Grid
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.plot(month, sales)
plt.grid()
plt.show()
```

```python
# Customize Line Chart
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.plot(month, sales)
plt.plot(
    month,
    sales,
    color="blue",
    marker="o",
    linestyle="-",
    linewidth=5,
    markersize=10
)

plt.show()
```

```python
# FIG SIZE
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.figure(figsize=(12, 4))
plt.plot(month, sales)
plt.show()
```

```python
# Chartname(title), x axis name and y axis name
import matplotlib.pyplot as plt
month = [1, 2, 3, 4, 5]
sales = [100, 200, 100, 300, 200]
plt.figure(figsize=(10, 3))
plt.title("Monthly Sales")
plt.xlabel("Sales")
plt.ylabel("Months")
plt.plot(month, sales)
plt.show()
```

# ADVANCED LINE CHART

```python
import matplotlib.pyplot as plt
x = ["Jan", "Feb", "Mar", "Apr", "May", "June"]
sales2025 = [100, 700, 500, 200, 300, 750]
sales2026 = [150, 600, 500, 300, 450, 650]
plt.figure(figsize=(10, 3))
plt.plot(x, sales2025, label="2025 Sales Data")
plt.plot(x, sales2026, label="2026 Sales Data")

plt.title("Year on Year Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# BAR CHART

```python
import matplotlib.pyplot as plt
month = ["Jan", "Feb", "Mar", "Apr","May"]
sales = [200, 300, 500, 600, 800]
plt.bar(month, sales)
plt.title("Monthly Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# Advance Bar Chart
import matplotlib.pyplot as plt
month = ["Jan", "Feb", "Mar", "Apr","May"]
sales2025 = [500, 500, 500, 800, 900]
sales2026 = [250, 350, 550, 650, 850]
plt.bar(month, sales2025, label="2025 Sales Data")
plt.bar(month, sales2026, label="2026 Sales Data")
plt.title("Monthly Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# HISTOGRAM

# PIE CHART

```python
import matplotlib.pyplot as plt
month = ["Jan", "Feb", "Mar", "Apr","May"]
sales = [500, 500, 500, 800, 900]
plt.pie(sales, labels=month, autopct="%1.1f%%", startangle=90)
plt.title("Sales Data")
plt.show()
```

# SCATTER PLOT

```python
import matplotlib.pyplot as plt
import numpy as np
hight = [np.random.randint(0, 100) for _ in range(10)]
print(hight)
width = [np.random.randint(0, 100) for _ in range(10)]
print(width)
plt.scatter(hight, width)
plt.title("Hight-Width Relationship")
plt.show()
```

# SUB-PLOT

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(10, 7))

# Line Chart
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
day_sales = [200, 160, 210, 300, 500]
plt.subplot(2, 3, 1)
plt.plot(days, day_sales)


# # Bar chart
month = ["Jan", "Feb", "Mar", "Apr", "May"]
month_sales = [2000, 1600, 2150, 3000, 5500]
plt.subplot(2, 3, 2)
plt.bar(month, month_sales)

# Histogram
year_sales = [
    20000, 16000, 21500, 30000, 55000,
    25000, 18000, 32000, 45000, 28000,
    35000, 22000, 50000, 40000, 30000
]
plt.subplot(2, 3, 3)
plt.hist(year_sales, bins=10, edgecolor="Red")

# Pie Chart
decade = [1980, 1990, 2000, 2010, 2020]
decade_sales = [200000, 160000, 215500, 300000, 450000]
plt.subplot(2, 3, 4)
plt.pie(decade_sales, labels=decade, autopct="%1.1f%%")


# Scatter Chart
offers_per= [np.random.randint(0, 100) for _ in range(0, 100)]
offers_date= [np.random.randint(0, 100) for _ in range(0, 100)]
plt.subplot(2, 3, 5)
plt.scatter(offers_per, offers_date)

plt.title("Master Sales Report")
plt.show()
```

# MATPLOTLIB USING DATAFRAME

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Months": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Sales": [2000, 2200, 1300, 3400, 5100, 6700, 4500, 1200, 5400, 2300, 9100, 7700]
}
df = pd.DataFrame(data)
print(df)
print("Data Types:",type(df))

# Create Bar chart with dataframe data
plt.figure(figsize=(10, 3))
plt.title("Monthy Sales")

plt.bar(df["Months"], df["Sales"])

plt.show()
```
