# LINE CHART

```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [1200, 900, 950, 1100, 1490, 1500, 1800]
plt.figure(figsize=(10, 4))
plt.plot(days, sales, color="purple", marker="o", linestyle=":", linewidth=3, markersize=15)
plt.title("Weekly Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

# ADVANCE LINE CHART

```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales1 = [1200, 900, 950, 1100, 1490, 1500, 1800]
sales2 = [1300, 800, 1050, 1200, 1490, 1500, 2000]
plt.figure(figsize=(10, 4))
plt.plot(days, sales1, label="Hyderabad Store Sales", marker="o", linestyle="--", color="green")
plt.plot(days, sales2, label="Banglore Store Sales", marker="o", linestyle="--", color="black")
plt.title("Weekly Sales")
plt.legend()
plt.grid()
plt.show()
```

# BAR CHART

```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [1200, 900, 950, 1100, 1490, 1500, 1800]
plt.figure(figsize=(8, 3))
plt.bar(days, sales, color="black")
plt.title("Weekly Sales Data")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

# ADVANCE BAR CHART

```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
banglore_sales = [1200, 900, 950, 1100, 1490, 1500, 1800]
noida_sales = [1200, 500, 950, 1100, 1490, 1200, 1700]
plt.figure(figsize=(8, 3))
plt.bar(days, banglore_sales, color="black", label="Banglore Sales Data")
plt.bar(days, noida_sales, color="purple", label="Noida Sales Data")
plt.title("Weekly Sales Data")
plt.legend()
plt.show()
```

# HISTOGRAM

```python
import matplotlib.pyplot as plt
age = [18, 19, 20, 21, 22, 22, 23, 24, 25, 25, 26, 27, 28, 30]
plt.title("Sales Data")
plt.hist(age, bins=3, edgecolor="black")
plt.show()
```

# PIE CHART

```python
import matplotlib.pyplot as plt
days = ["Mon", "Wed", "Fri", "Sun"]
sales = [1200, 900, 1150, 1700]
plt.title("Sales Data")
plt.pie(sales, labels=days, autopct="%1.2f%%", startangle=270)
plt.show()
```

# SCATTER PLOT

```python
import matplotlib.pyplot as plt
import numpy as np
x = [np.random.randint(0, 100) for _ in range(0, 100)]
y = [np.random.randint(100, 1000) for _ in range(0, 100)]
plt.title("Random Data in Scatter Plot")
plt.scatter(x, y, color="purple")
plt.show()
```

# SUB-PLOT

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(15, 7))
plt.title("Sales Data")

# Line Chart
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [200, 160, 210, 300, 500]
sales1 = [201, 190, 219, 300, 500]

# Line chart
plt.subplot(3, 2, 1)
plt.plot(days, sales)

# Bar chart
plt.subplot(3, 2, 2)
plt.bar(days, sales)

# Histogram
plt.subplot(3, 2, 3)
plt.hist(sales, bins=3)


# Pie chart
plt.subplot(3, 2, 4)
plt.pie(sales, labels=days)


# Scatter
plt.subplot(3, 2, 3)
plt.scatter(sales, sales1)


plt.show()

```

# MATPLOTLIB WITH PANDAS


```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "days": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "sales": [5200, 4300, 2300, 700, 4500]
}
# convert in dataframe
df = pd.DataFrame(data)
print(df)

# Craete graph using matplotlib
plt.plot(df["days"], df["sales"], marker="o", linestyle="--", color="red")
plt.title("Weely sales data")
plt.xlabel("Days")
plt.ylabel("sales")
plt.show()
```
